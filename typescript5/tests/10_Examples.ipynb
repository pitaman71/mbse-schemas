# Suite EX — Examples

Every example under `src/Examples` runs clean in its own process, as a program would. The examples are self-checking; a non-zero exit is a failure.

```typescript
import { spawnSync } from "node:child_process";
import { readdirSync } from "node:fs";

import { assert, same } from "./support.js";

const EXAMPLES = readdirSync("../src/Examples").filter((n) => /^[A-Z].*\.ts$/.test(n)).map((n) => n.replace(/\.ts$/, "")).sort();
assert(same(EXAMPLES, ["AddressBook", "FamilyTree", "SchemaBuilders", "Serialization", "TextEncodings", "University"]), EXAMPLES.join());
```

## EX-01 · Each example exits cleanly

```typescript
{
  const failures: Record<string, string> = {};
  for (const name of EXAMPLES) {
    const result = spawnSync("npx", ["tsx", `../src/Examples/${name}.ts`], { encoding: "utf8", timeout: 300_000 });
    if (result.status !== 0) failures[name] = (result.stderr || "").slice(-2000);
  }
  assert(Object.keys(failures).length === 0, JSON.stringify(failures));
}
```
