# Suite EX — Examples

Every example under `mbse_schemas/Examples` runs clean in its own process (the proxy registry is global, so they cannot
share one). The examples are self-checking; a non-zero exit is a failure.

```python
import subprocess
import sys
from pathlib import Path

import mbse_schemas.Examples as examples_package

EXAMPLES = sorted(p.stem for p in Path(list(examples_package.__path__)[0]).glob("[A-Z]*.py"))
assert EXAMPLES == ["AddressBook", "FamilyTree", "SchemaRules", "Serialization", "TextEncodings", "University"], EXAMPLES
```

## EX-01 · Each example exits cleanly

```python
failures = {}
for name in EXAMPLES:
    result = subprocess.run([sys.executable, "-m", f"mbse_schemas.Examples.{name}"], capture_output=True, text=True,
                            timeout=300)
    if result.returncode != 0:
        failures[name] = result.stderr[-2000:]
assert failures == {}, failures
```
